**Problem 8** *(Computational complexity and heaps, 12 marks)*

Consider the following barebones BFS implementation, where the graph has $|\calV|$ vertices and $|\calE|$ edges:

```python
def bfs_barebones(graph, start):
    seen = {start}                        # line 1
    frontier = Queue()                    # line 2
    frontier.put(start)                   # line 3
    while not frontier.empty():           # line 4
        m = frontier.get()                # line 5
        for neighbor in graph.get(m, []):  # line 6
            if neighbor not in seen:         # line 7
                seen.add(neighbor)              # line 8
                frontier.put(neighbor)          # line 9
```

a. *(6 marks)* Annotate each line (1-9) with its individual computational complexity in big-O notation, and use these to show that the overall complexity of BFS is $O(|\calV| + |\calE|)$.

b. *(6 marks)* State the max-heap property relating a node to its parent, and give the array-index formulas for `Parent(i)`, `LeftChild(i)`, `RightChild(i)` used in this course's notes for a 1-indexed array-based heap. Then state the time complexity of `Heap-Insert`, `Heap-Extract-Max`, and `Build-Heap` (heapifying an entire array of $n$ elements) in terms of $n$.
